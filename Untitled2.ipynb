```python
import random

# In-memory storage for routes and bookings
routes = [
    {
        "id": "R001",
        "origin": "Karachi",
        "destination": "Lahore",
        "total_seats": 40,
        "available_seats": 40,
        "fare": 3500
    },
    {
        "id": "R002",
        "origin": "Karachi",
        "destination": "Islamabad",
        "total_seats": 35,
        "available_seats": 35,
        "fare": 5000
    },
    {
        "id": "R003",
        "origin": "Lahore",
        "destination": "Islamabad",
        "total_seats": 30,
        "available_seats": 30,
        "fare": 1800
    },
    {
        "id": "R004",
        "origin": "Multan",
        "destination": "Karachi",
        "total_seats": 40,
        "available_seats": 40,
        "fare": 3000
    }
]

bookings = {}  # Key: booking_id, Value: booking_details dict

def display_menu():
    print("\n" + "="*45)
    print("   SMART BUS TICKET BOOKING & RESERVATION SYSTEM")
    print("="*45)
    print("1. Browse Routes")
    print("2. Search Journeys")
    print("3. Reserve Seats")
    print("4. Inspect Bookings")
    print("5. Cancel Reservation")
    print("6. Check Seat Availability")
    print("7. View Reports")
    print("8. Exit")
    print("="*45)

def browse_routes():
    print("\n--- Available Routes ---")
    print(f"{'Route ID':<10} {'Origin':<15} {'Destination':<15} {'Seats Avail':<12} {'Fare (PKRs)':<10}")
    print("-" * 65)
    for r in routes:
        print(f"{r['id']:<10} {r['origin']:<15} {r['destination']:<15} {r['available_seats']:<12} {r['fare']:<10}")

def search_journeys():
    print("\n--- Search Journeys ---")
    origin = input("Enter Departure City (Origin): ").strip().title()
    destination = input("Enter Arrival City (Destination): ").strip().title()

    found = False
    print(f"\nSearching journeys from {origin} to {destination}...")
    print(f"{'Route ID':<10} {'Origin':<15} {'Destination':<15} {'Seats Avail':<12} {'Fare (PKRs)':<10}")
    print("-" * 65)
    for r in routes:
        if r['origin'].lower() == origin.lower() and r['destination'].lower() == destination.lower():
            print(f"{r['id']:<10} {r['origin']:<15} {r['destination']:<15} {r['available_seats']:<12} {r['fare']:<10}")
            found = True

    if not found:
        print("No routes found matching your search criteria.")

def reserve_seats():
    print("\n--- Reserve Seats ---")
    browse_routes()
    route_id = input("\nEnter Route ID to book: ").strip().upper()

    selected_route = None
    for r in routes:
        if r['id'] == route_id:
            selected_route = r
            break

    if not selected_route:
        print("Error: Invalid Route ID.")
        return

    try:
        seats_to_book = int(input(f"Enter number of seats to reserve (Available: {selected_route['available_seats']}): "))
    except ValueError:
        print("Error: Please enter a valid numeric value for seats.")
        return

    if seats_to_book <= 0:
        print("Error: Number of seats must be greater than zero.")
        return

    if seats_to_book > selected_route['available_seats']:
        print("Error: Not enough available seats on this route.")
        return

    passenger_name = input("Enter Passenger Name: ").strip()
    if not passenger_name:
        print("Error: Passenger name cannot be empty.")
        return

    # Generate unique booking ID
    booking_id = f"BK-{random.randint(1000, 9999)}"
    while booking_id in bookings:
        booking_id = f"BK-{random.randint(1000, 9999)}"

    total_cost = seats_to_book * selected_route['fare']

    # Update route available seats
    selected_route['available_seats'] -= seats_to_book

    # Save booking
    bookings[booking_id] = {
        "booking_id": booking_id,
        "route_id": selected_route['id'],
        "origin": selected_route['origin'],
        "destination": selected_route['destination'],
        "passenger_name": passenger_name,
        "seats": seats_to_book,
        "total_cost": total_cost,
        "status": "Confirmed"
    }

    print("\nSUCCESS: Reservation Confirmed!")
    print(f"Booking ID   : {booking_id}")
    print(f"Passenger    : {passenger_name}")
    print(f"Route        : {selected_route['origin']} -> {selected_route['destination']}")
    print(f"Seats Booked : {seats_to_book}")
    print(f"Total Fare   : PKRs {total_cost}")

def inspect_bookings():
    print("\n--- Inspect Bookings ---")
    if not bookings:
        print("No bookings found in the system.")
        return

    print(f"{'Booking ID':<12} {'Passenger':<18} {'Route':<22} {'Seats':<8} {'Status':<12} {'Total (PKRs)':<10}")
    print("-" * 85)
    for b_id, b in bookings.items():
        route_str = f"{b['origin']}-{b['destination']}"
        print(f"{b_id:<12} {b['passenger_name']:<18} {route_str:<22} {b['seats']:<8} {b['status']:<12} {b['total_cost']:<10}")

def cancel_reservation():
    print("\n--- Cancel Reservation ---")
    booking_id = input("Enter Booking ID to cancel: ").strip().upper()

    if booking_id not in bookings:
        print("Error: Booking ID not found.")
        return

    booking = bookings[booking_id]

    if booking['status'] == "Cancelled":
        print("Notice: This booking has already been cancelled.")
        return

    # Restore seats to the corresponding route
    for r in routes:
        if r['id'] == booking['route_id']:
            r['available_seats'] += booking['seats']
            break

    booking['status'] = "Cancelled"
    print(f"SUCCESS: Booking {booking_id} has been successfully cancelled. Seats have been restored.")

def check_availability():
    print("\n--- Check Seat Availability ---")
    browse_routes()

def view_reports():
    print("\n--- System Reports & Summary ---")
    if not bookings:
        print("No transactions recorded yet.")
        return

    total_bookings = len(bookings)
    active_bookings = sum(1 for b in bookings.values() if b['status'] == "Confirmed")
    cancelled_bookings = sum(1 for b in bookings.values() if b['status'] == "Cancelled")

    # Revenue excludes cancelled bookings
    confirmed_revenue = sum(b['total_cost'] for b in bookings.values() if b['status'] == "Confirmed")

    print(f"Total Bookings Attempted : {total_bookings}")
    print(f"Active Confirmed Bookings: {active_bookings}")
    print(f"Cancelled Bookings       : {cancelled_bookings}")
    print(f"Confirmed Revenue        : PKRs {confirmed_revenue}")

def main():
    while True:
        display_menu()
        choice = input("\nEnter your choice (1-8): ").strip()

        if choice == '1':
            browse_routes()
        elif choice == '2':
            search_journeys()
        elif choice == '3':
            reserve_seats()
        elif choice == '4':
            inspect_bookings()
        elif choice == '5':
            cancel_reservation()
        elif choice == '6':
            check_availability()
        elif choice == '7':
            view_reports()
        elif choice == '8':
            print("\nThank you for using Smart Bus Ticket Booking System. Goodbye!")
            break
        else:
            print("\nInvalid choice! Please enter a number between 1 and 8.")

if __name__ == "__main__":
    main()
```
